```python
import requests

url = "http://localhost:8000/aiAsk/canIEat"

headers = {
    "Content-Type": "application/json",
    "Authorization": "Bearer abc123"
}


#요청 body 예시 데이터
#food_name=필수, 나머지는 선택, manufacture의 기본값=오늘
data = {
    "food_name": "초코우유",
    "manufacture_date": "2026-05-29",
    "expiration_date": "2026-06-18",
    "allergies": ["우유", "두유"],
    "ingredients": ["밀가루", "설탕", "혼합분유"]
}

response = requests.post(
    url,
    json=data
)

print(response.text)
```
